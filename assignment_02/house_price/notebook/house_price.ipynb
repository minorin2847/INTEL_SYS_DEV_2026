## 5. Data-quality analysis

Every problem found — impossible values, missingness, constant columns, leakage,
re-scrapes — goes into one *issue → count → planned action* table.


````python
PRICE_LO, PRICE_HI = 0, 200_000        # million VND  (~0 .. ~200 tỷ)
AREA_LO, AREA_HI   = 0, 10_000         # m^2

rows = []
rows.append(['Price', 'value <= 0 (impossible)', int((df['Price'] <= PRICE_LO).sum()), 'drop row'])
rows.append(['Price', f'value > {PRICE_HI} (implausible outlier)', int((df['Price'] > PRICE_HI).sum()), 'drop row'])
rows.append(['Area',  'value <= 0 (impossible)', int((df['Area'] <= AREA_LO).sum()), 'drop row'])
rows.append(['Area',  f'value > {AREA_HI} (implausible outlier)', int((df['Area'] > AREA_HI).sum()), 'drop row'])
rows.append(['Area',  'missing', int(df['Area'].isna().sum()), 'drop row (target-critical size)'])
for c in ['Bedrooms', 'Bathrooms', 'Floors']:
    rows.append([c, 'missing', int(df[c].isna().sum()), 'median impute + missing-indicator flag'])
for c in ['Width', 'Length', 'Alley Width']:
    rows.append([c, 'missing', int(df[c].isna().sum()), 'median impute (in pipeline)'])
for c in ['Position', 'Direction', 'Road Type']:
    rows.append([c, 'missing', int(df[c].isna().sum()), "fill 'Unknown' category"])
rows.append(['VIP Account', 'constant column', int(df['VIP Account'].nunique()), 'drop column'])
rows.append(['Listing ID', 'duplicate ids', int(df['Listing ID'].duplicated().sum()), 'drop dupes by id (section 7)'])
rows.append(['Title', 'free text — embeds Price', df['Title'].notna().sum(), 'DROP (target leakage)'])
rows.append(['Description', 'free text', int(df['Description'].isna().sum()), 'drop column (not used for tabular model)'])
quality = pd.DataFrame(rows, columns=['column', 'issue', 'count', 'planned action'])
quality
````

The quality problems cluster in a handful of columns. There are **impossible
values**: 611 listings with `Price ≤ 0`, roughly 10,600 above 200,000 million VND, and
a handful of `Area ≤ 0` or `Area >` 10,000 m² — data-entry noise, so those rows are
dropped. There is **structural missingness**: `Bathrooms` / `Floors` / `Bedrooms` /
`Alley Width` / `Direction` / `Road Type` sit 70–84% empty, which calls for imputing
and flagging rather than dropping rows — dropping would erase most of the data.
`VIP Account` holds a single constant value, so the column is dropped outright.
`Listing ID` repeats 19,512 times — the same advert re-scraped — handled in section 7.
And `Title` states the price directly, so it is dropped to avoid leaking the target.
Each of these actions is carried out in sections 6–9 and 13–15.

## 7. Duplicate analysis

Check both for exact duplicate rows and for repeated `Listing ID` values — the same
advert picked up by the crawler on more than one day.


````python
n_before = len(df)
print('exact duplicate rows        :', df.duplicated().sum())
print('duplicate Listing ID values :', df['Listing ID'].duplicated().sum())

# Listing ID is a natural key — the same advert re-scraped on different days.
df = df.drop_duplicates(subset='Listing ID', keep='last').reset_index(drop=True)
print(f'rows: {n_before} -> {len(df)}  (removed {n_before - len(df)} re-scrapes)')
````

There are no exact duplicate rows, but 19,512 `Listing ID` values repeat — the

become 216,714. Doing this **before** the section-14 train/test split is what keeps
the same property from landing on both sides and inflating the score.


## 8. Invalid-value analysis

Apply the section-5 domain bounds (`0 < Price ≤ 200,000` million VND,
`0 < Area ≤ 10,000` m²) and drop whatever violates them.


````python
checks = {
    'Price <= 0'            : (df['Price'] <= PRICE_LO).sum(),
    f'Price > {PRICE_HI}'   : (df['Price'] > PRICE_HI).sum(),
    'Area <= 0'             : (df['Area'] <= AREA_LO).sum(),
    f'Area > {AREA_HI}'     : (df['Area'] > AREA_HI).sum(),
    'Bedrooms < 0'          : (df['Bedrooms'] < 0).sum(),
    'Floors < 0'            : (df['Floors'] < 0).sum(),
}
print(pd.Series(checks, name='violations'))

n_before = len(df)
df = df[(df['Price'] > PRICE_LO) & (df['Price'] <= PRICE_HI)]
df = df[(df['Area'].notna()) & (df['Area'] > AREA_LO) & (df['Area'] <= AREA_HI)]
df = df.reset_index(drop=True)
print(f'\nrows after domain filter: {n_before} -> {len(df)}  ({len(df)/n_before*100:.1f}% kept)')
````

`Price = 0` / `9e12` and `Area = −6` / `1e18` are recording errors rather than
signal, so those rows are dropped outright — capping them would only keep meaningless
values around. The bounds `0 < Price ≤ 200,000` million VND and `0 < Area ≤ 10,000` m²
together keep about 93% of listings.


## 9. Outlier analysis

IQR-rule counts and box plots on the cleaned numeric columns, to decide between
keeping and capping the long right tails.


````python
num_cols_raw = ['Price', 'Area', 'Width', 'Length', 'Bedrooms', 'Bathrooms', 'Floors', 'Alley Width']
Q1, Q3 = df[num_cols_raw].quantile(0.25), df[num_cols_raw].quantile(0.75)
IQR = Q3 - Q1
iqr_outliers = ((df[num_cols_raw] < Q1 - 1.5*IQR) | (df[num_cols_raw] > Q3 + 1.5*IQR)).sum()
print('IQR-rule outlier counts:')
print(iqr_outliers)

fig, ax = plt.subplots(2, 4, figsize=(15, 6))
for a, c in zip(ax.ravel(), num_cols_raw):
    df.boxplot(column=c, ax=a); a.set_title(c)
plt.tight_layout(); plt.show()
````

`Price`, `Area`, `Width` and `Length` all have long right tails, but with the

rows stay; the model is trained on **`log1p(Price)`** (section 12) to compress the
tail, and `Area` / `Width` / `Length` / `Alley Width` are capped at the 99th
percentile (section 13).

### 10.2 Relationships

Each feature is checked against the target first (10.2.1–10.2.3), then the
categorical-strength view (η², 10.2.5), then two further views drill into the
strongest categoricals: district-level location (10.2.6) and the broker-vs-owner
listing channel (10.2.7).


````python
# correlation ratio (eta^2): share of variance in `values` explained by a categorical grouping
def correlation_ratio(categories, values):
    cats = categories.astype('category').cat.codes.values
    y = values.values
    ss_total = ((y - y.mean()) ** 2).sum()
    ss_between = sum(len(y[cats == k]) * (y[cats == k].mean() - y.mean()) ** 2
                     for k in np.unique(cats))
    return ss_between / ss_total if ss_total > 0 else 0.0
````

````python
# ---- 10.2.1 · log Price vs log Area, all property types pooled ----
samp = df.sample(min(8000, len(df)), random_state=RANDOM_SEED)
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(samp['area_log'], samp['price_log'], s=6, alpha=0.3)
ax.set_title('10.2.1 · log Price vs log Area (pooled)')
ax.set_xlabel('log1p(Area)'); ax.set_ylabel('log1p(Price)')
plt.tight_layout(); plt.show()

print('pooled Pearson r(log Area, log Price) =',
      round(df['area_log'].corr(df['price_log']), 3))
````

The pooled cloud shows only a **weak** positive trend — Pearson r ≈ 0.02 — with
very wide vertical spread at every area value. Taken across all listings at once, area
barely predicts price; section 10.2.3 and 10.3.4 show this is misleading, because the
land-versus-built mix cancels the effect out. Area on its own is not a usable linear
predictor — it needs the property-type context to mean anything.


````python
# ---- 10.2.2 · price level by property type and by province ----
fig, ax = plt.subplots(1, 2, figsize=(15, 4.5))
df.groupby('Property Type')['Price'].median().sort_values().plot(kind='barh', ax=ax[0])
ax[0].set_title('10.2.2a · median Price by Property Type'); ax[0].set_xlabel('median Price (M VND)')

top_prov = df['Province'].value_counts().head(15).index
(df[df['Province'].isin(top_prov)]
   .groupby('Province')['price_per_m2'].median().sort_values()
   .plot(kind='barh', ax=ax[1]))
ax[1].set_title('10.2.2b · median Price/m² — top-15 provinces'); ax[1].set_xlabel('M VND / m²')
plt.tight_layout(); plt.show()
````

Median price differs several-fold across property types, and median price per
m² varies roughly five-fold across provinces, with HCMC, Hà Nội and Đà Nẵng highest
and rural provinces lowest. Property type and province each set the *level* that price
sits at; area then scales the price within that level. Both are one-hot encoded in
section 15, and both are expected to come out among the strongest features — a claim
confirmed later by 10.2.5's η² figures and section 20's importances.


````python
# ---- 10.2.3 · the same relationship, split by property type ----
fig, ax = plt.subplots(1, 2, figsize=(16, 4.5))
for pt in df['Property Type'].value_counts().head(4).index:
    d = df[df['Property Type'] == pt].sample(
        min(2000, (df['Property Type'] == pt).sum()), random_state=RANDOM_SEED)
    ax[0].scatter(d['area_log'], d['price_log'], s=6, alpha=0.25, label=pt)
ax[0].legend(fontsize=8)
ax[0].set_title('10.2.3a · log Price vs log Area, by Property Type')
ax[0].set_xlabel('log1p(Area)'); ax[0].set_ylabel('log1p(Price)')

within = (df.groupby('Property Type')
            .apply(lambda g: g['area_log'].corr(g['price_log']))
            .sort_values())
within.plot(kind='barh', ax=ax[1], color='#55A868')
for i, v in enumerate(within.values):
    ax[1].text(v + (0.01 if v >= 0 else -0.01), i, f'{v:.2f}', va='center',
               ha='left' if v >= 0 else 'right', fontsize=9)
ax[1].axvline(0, color='k', lw=0.8)
ax[1].set_title('10.2.3b · corr(log Area, log Price) WITHIN each type')
ax[1].set_xlabel('Pearson r')
plt.tight_layout(); plt.show()
````

Split by property type, the area-to-price slope turns out positive for the built
types (`Nhà riêng`, `Căn hộ chung cư`, `Văn phòng`), close to flat for `Đất`, and
roughly zero once everything is pooled together (10.2.1). Area should therefore stay
in the model, but only in combination with `Property Type` — something tree ensembles
already do natively, as the per-type slopes in 10.3.4 confirm.


````python
# ---- 10.2.5 · categorical → log Price strength (correlation ratio η²) ----
cat_cols = ['Property Type', 'Province', 'Position', 'Direction', 'Road Type', 'Agent Role']
eta = pd.Series({c: correlation_ratio(df[c].fillna('Unknown'), df['price_log'])
                for c in cat_cols}).sort_values()

fig, ax = plt.subplots(figsize=(8, 4))
eta.plot(kind='barh', ax=ax, color='#4C72B0')
for i, v in enumerate(eta.values):
    ax.text(v + 0.005, i, f'{v:.3f}', va='center', fontsize=9)
ax.set_xlim(0, max(0.3, eta.max() * 1.25))
ax.set_title('10.2.5 · categorical → log Price strength (η², variance explained)')
ax.set_xlabel('η²  (0 = no effect, 1 = fully explains price)')
plt.tight_layout(); plt.show()

print(eta.sort_values(ascending=False).round(3).to_string())
````

`Province` reaches η² ≈ 0.24 and `Property Type` ≈ 0.12 — comparable to r², each
explains more price variance on its own than any numeric feature does, while
`Direction`, `Position`, `Road Type` and `Agent Role` all sit under 0.05. `Property
Type` and `Province` / `district` are one-hot encoded; the low-η² categoricals are the
first candidates to drop if the feature count `d` ever needs to shrink.


````python
# ---- 10.2.6 · district-level price (the engineered `district`, one level below Province) ----
# `district` is built in section 13; parse it here just for EDA (Quận/Huyện/TP token, else ward).
def _district(loc):
    if not isinstance(loc, str):
        return 'Unknown'
    parts = [p.strip() for p in loc.replace('(Mới)', '').split(',') if p.strip()]
    d = next((p for p in parts if p.startswith(('Quận', 'Huyện', 'Thị xã', 'Thành phố', 'TP.', 'TP '))), None)
    return d or next((p for p in parts if p.startswith(('Phường', 'Xã', 'Thị trấn', 'Đặc khu'))), 'Unknown')

_dist = df['Location'].apply(_district)
_d = pd.DataFrame({'district': _dist, 'Price': df['Price'], 'ppm2': df['price_per_m2']})
_g = (_d[_d['district'].ne('Unknown')]
        .groupby('district')
        .agg(n=('Price', 'size'), med_price=('Price', 'median'), med_ppm2=('ppm2', 'median')))
_g = _g[_g['n'] >= 100]                              # keep districts with a usable sample
print(f'{_dist.nunique():,} distinct district/ward tokens; {len(_g)} with n >= 100 listings')

fig, ax = plt.subplots(1, 2, figsize=(15, 5))
_g.sort_values('med_price').tail(15)['med_price'].plot(kind='barh', ax=ax[0], color='#4C72B0')
ax[0].set_title('10.2.6a · top-15 districts by median Price'); ax[0].set_xlabel('median Price (M VND)')
_g.sort_values('med_ppm2').tail(15)['med_ppm2'].plot(kind='barh', ax=ax[1], color='#55A868')
ax[1].set_title('10.2.6b · top-15 districts by median Price / m²'); ax[1].set_xlabel('M VND / m²')
plt.tight_layout(); plt.show()

print('\nmost / least expensive districts (median Price, M VND):')
print(_g.sort_values('med_price', ascending=False).head(5).round(1).to_string())
print('...')
print(_g.sort_values('med_price').head(5).round(1).to_string())
````

District pulls the price level far harder than province alone: central Hà Nội /
HCMC districts (`Quận Hoàn Kiếm` at a median around 49,000 million VND, `Quận 1`
around 33,000, `Quận 3` / `Tây Hồ` / `Ba Đình` around 25,000) sit roughly 100–200×
above the cheapest rural districts (`Thị xã Bình Long`, `Xã Tân Hưng`,
`Huyện Hớn Quản`, around 200–300 million VND). The gap in price per m² is even wider —
roughly 600 versus 0.5 million VND/m². Within a single province, district explains a
large share of the remaining spread: a listing in `Quận 1` and one in an outer HCMC
district both carry `Province = ho-chi-minh` yet sit an order of magnitude apart.
`district` (parsed in section 13) is kept as a one-hot feature alongside `Province`;
rare districts below `min_frequency` fold into an `infrequent` bucket, which section
20 later flags as a source of error. Ward-level location (~4,600 levels) stays
dropped.


````python
# ---- 10.2.7 · listing channel: broker (Môi giới) vs owner (Chính chủ) ----
_c = pd.DataFrame({'role': df['Agent Role'], 'Price': df['Price'],
                   'ppm2': df['price_per_m2'], 'Area': df['Area'],
                   'Property Type': df['Property Type']})
role_tbl = _c.groupby('role').agg(n=('Price', 'size'), share_pct=('Price', lambda s: 100*len(s)/len(_c)),
                                  med_price=('Price', 'median'), med_ppm2=('ppm2', 'median'),
                                  med_area=('Area', 'median'))
print(role_tbl.round(1).to_string())

top4 = _c['Property Type'].value_counts().head(4).index
piv = (_c[_c['Property Type'].isin(top4)]
         .pivot_table('ppm2', 'Property Type', 'role', 'median'))

fig, ax = plt.subplots(1, 2, figsize=(14, 4.5))
role_tbl['med_ppm2'].plot(kind='bar', ax=ax[0], color=['#C44E52', '#4C72B0'])
ax[0].set_title('10.2.7a · median Price / m² by listing channel')
ax[0].set_ylabel('M VND / m²'); ax[0].set_xlabel(''); ax[0].tick_params(axis='x', rotation=0)
piv.plot(kind='barh', ax=ax[1])
ax[1].set_title('10.2.7b · median Price / m² — channel × property type')
ax[1].set_xlabel('M VND / m²'); ax[1].legend(title='')
plt.tight_layout(); plt.show()
````

About 91% of listings go through a broker (`Môi giới`) and 9% are listed by the

## 13. Feature engineering

Parse `district` out of `Location`, add the `*_missing` indicator flags, and cap
`Area` / `Width` / `Length` / `Alley Width` at the 99th percentile.


````python
def parse_district(loc):
    """'Đường X,Phường Y,Quận Z,Tỉnh(Mới)' -> district.
    Prefer a Quận/Huyện/Thị xã/Thành phố token; fall back to the ward
    (Phường/Xã/Thị trấn/Đặc khu); 'Unknown' if neither is present.
    Ward itself is NOT used as a feature: ~4,600 levels is mostly noise."""
    if not isinstance(loc, str):
        return 'Unknown'
    parts = [p.strip() for p in loc.replace('(Mới)', '').split(',') if p.strip()]
    dist = next((p for p in parts if p.startswith(('Quận', 'Huyện', 'Thị xã', 'Thành phố', 'TP.', 'TP '))), None)
    if dist:
        return dist
    return next((p for p in parts if p.startswith(('Phường', 'Xã', 'Thị trấn', 'Đặc khu'))), 'Unknown')


def engineer(frame):
    out = frame.copy()
    # 1. parsed location (district level only)
    out['district'] = out['Location'].apply(parse_district)
    # 2. missing-indicator flags for the >50%-missing counts
    for c in ['Bedrooms', 'Bathrooms', 'Floors']:
        out[c + '_missing'] = out[c].isna().astype(int)
    # 3. cap extreme size (99th pct) so distance/linear models aren't dominated
    for c in ['Area', 'Width', 'Length', 'Alley Width']:
        cap = out[c].quantile(0.99)
        out[c] = out[c].clip(upper=cap)
    # 4. fill categorical NaN with an explicit level
    for c in ['Position', 'Direction', 'Road Type']:
        out[c] = out[c].fillna('Unknown')
    return out


df_eng = engineer(df)

NUM_FEATURES = ['Area', 'Width', 'Length', 'Bedrooms', 'Bathrooms', 'Floors',
                'Alley Width', 'Agent Listing Count',
                'Bedrooms_missing', 'Bathrooms_missing', 'Floors_missing']
CAT_FEATURES = ['Property Type', 'Position', 'Direction', 'Road Type',
                'Province', 'Agent Role', 'district']
FEATURES = NUM_FEATURES + CAT_FEATURES

print('final raw feature count :', len(FEATURES))
print('numeric  :', NUM_FEATURES)
print('categoric:', CAT_FEATURES)
print('district levels        :', df_eng['district'].nunique(),
      '(rare ones folded by min_frequency in section 15)')
df_eng[FEATURES].head(3)
````

**Engineered features.**

| Feature | Purpose |
|---|---|
| `district` (from `Location`) | admin-level location signal; `ward` (~4,600 levels) dropped as noise. |
| `Bedrooms_missing` / `Bathrooms_missing` / `Floors_missing` | encode "was this recorded?" (>70% missing, correlates with type). |
| 99th-pct cap on `Area` / `Width` / `Length` / `Alley Width` | bound the influence of a few huge lots. |

Every categorical is one-hot encoded, since none carries a natural order;
`OneHotEncoder(handle_unknown='ignore', min_frequency=50)` keeps the resulting column
count in check. The fitted pipeline (section 15) ends up with **`d = 404`**.


## 14. Train / validation / test split

The split happens **before** any `.fit()` call — roughly 70 / 15 / 15,
`random_state = 42`. The section-7 `Listing ID` dedupe already keeps any one property
off both sides.


````python
from sklearn.model_selection import train_test_split

X_all = df_eng[FEATURES].copy()
y_all = np.log1p(df_eng['Price'].values)          # model target
y_all_vnd = df_eng['Price'].values                # kept for reporting on the real scale

X_train, X_tmp, y_train, y_tmp, _, y_tmp_vnd = train_test_split(
    X_all, y_all, y_all_vnd, test_size=0.30, random_state=RANDOM_SEED)
X_val, X_test, y_val, y_test, y_val_vnd, y_test_vnd = train_test_split(
    X_tmp, y_tmp, y_tmp_vnd, test_size=0.50, random_state=RANDOM_SEED)

print(f'train: {X_train.shape}   val: {X_val.shape}   test: {X_test.shape}   (~70/15/15)')
````

Preprocessing is fitted on `X_train` only. If the imputer's medians, the
scaler's statistics or the one-hot vocabulary had seen validation or test rows, the
reported error would look better than it would on genuinely new data. The split
happens before any `.fit()` call, the section-7 `Listing ID` dedupe keeps a single
property from landing on both sides, and `random_state` is fixed throughout.


## 15. Preprocessing pipeline

One `ColumnTransformer`: median-impute and standardise the numeric columns,
`'Unknown'`-impute and one-hot the categoricals — fitted on the training split only.


````python
numeric_pipe = Pipeline([
    ('impute', SimpleImputer(strategy='median')),
    ('scale', StandardScaler()),
])
categorical_pipe = Pipeline([
    ('impute', SimpleImputer(strategy='constant', fill_value='Unknown')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', min_frequency=50, sparse_output=True)),
])

preprocessor = ColumnTransformer([
    ('num', numeric_pipe, NUM_FEATURES),
    ('cat', categorical_pipe, CAT_FEATURES),
])

preprocessor.fit(X_train)                    # <-- fitted on TRAIN ONLY
X_train_pp = preprocessor.transform(X_train)
D_FEATURES = X_train_pp.shape[1]
print('preprocessed train shape:', X_train_pp.shape, '  # d =', D_FEATURES)
print('(numeric', len(NUM_FEATURES), '+ one-hot', D_FEATURES - len(NUM_FEATURES),
      'indicator columns, rare levels folded into an <infrequent> column)')
````

The `ColumnTransformer` turns the 18 raw feature columns into **`d = 404`**: 11
numeric columns, median-imputed and standardised, plus 393 one-hot indicator columns.
`OneHotEncoder(min_frequency=50)` folds every rare category level — mostly small
`district` and `Province` values — into a single `infrequent` column, keeping `d`
bounded, which is why section 20 lists that `infrequent` bucket as a source of error.
`preprocessor.fit(X_train)` runs on the **training split only** — the imputer's
medians, the scaler's mean/std and the one-hot vocabulary never see validation or
test rows.


## 16. Baseline model

A `DummyRegressor` that predicts the median log-price, plus the scoring helper —
every later model is measured against this floor.


````python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

LOG_CLIP = np.log1p(1_000_000)   # cap predictions at 1,000,000 million VND before expm1

def _to_vnd(y_log):
    """Invert the log1p target, guarding against overflow from wild linear extrapolation."""
    return np.expm1(np.clip(y_log, None, LOG_CLIP))

def regression_scores(y_true_log, y_pred_log):
    """Scores reported on the real VND scale (million VND)."""
    yt, yp = _to_vnd(y_true_log), _to_vnd(y_pred_log)
    mae = mean_absolute_error(yt, yp)
    mse = mean_squared_error(yt, yp)
    rmse = np.sqrt(mse)
    r2 = r2_score(yt, yp)
    mape = np.mean(np.abs((yt - yp) / np.clip(yt, 1e-6, None))) * 100
    return {'MAE': mae, 'MSE': mse, 'RMSE': rmse, 'R2': r2, 'MAPE_%': mape}

base = Pipeline([('prep', preprocessor), ('reg', DummyRegressor(strategy='median'))])
base.fit(X_train, y_train)
base_scores = regression_scores(y_val, base.predict(X_val))
print('baseline (predict median log-price):')
for k, v in base_scores.items():
    print(f'  {k:7s}: {v:,.3f}')
````

## 17. Model training

Five regression models go head to head: Linear Regression, Ridge, Decision Tree,
Random Forest and Gradient Boosting (XGBoost). All five share the same `preprocessor`
and the same `X_train` / `y_train` on the log-transformed target.


````python
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
try:
    from xgboost import XGBRegressor
    _HAS_XGB = True
except ImportError:
    from sklearn.ensemble import GradientBoostingRegressor
    _HAS_XGB = False

# Train all models on the SAME sample so the comparison is fair and the notebook
# runs in a few minutes. Raise / remove TRAIN_SAMPLE for a final run.
TRAIN_SAMPLE = 50_000
if len(X_train) > TRAIN_SAMPLE:
    samp_idx = np.random.RandomState(RANDOM_SEED).choice(len(X_train), TRAIN_SAMPLE, replace=False)
    X_tr_s, y_tr_s = X_train.iloc[samp_idx], y_train[samp_idx]
else:
    X_tr_s, y_tr_s = X_train, y_train
print(f'training sample: {X_tr_s.shape[0]:,} rows')

models = {
    'LinearRegression': LinearRegression(),
    'Ridge'           : Ridge(alpha=10.0, random_state=RANDOM_SEED),
    'DecisionTree'    : DecisionTreeRegressor(max_depth=12, min_samples_leaf=25, random_state=RANDOM_SEED),
    'RandomForest'    : RandomForestRegressor(n_estimators=120, max_depth=18, min_samples_leaf=10,
                                              n_jobs=-1, random_state=RANDOM_SEED),
}
if _HAS_XGB:
    models['XGBoost'] = XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=6,
                                     subsample=0.8, colsample_bytree=0.8, tree_method='hist',
                                     objective='reg:squarederror', random_state=RANDOM_SEED, n_jobs=-1)
else:
    models['GradientBoosting'] = GradientBoostingRegressor(random_state=RANDOM_SEED)

fitted, timings = {}, {}
for name, reg in models.items():
    pipe = Pipeline([('prep', preprocessor), ('reg', reg)])
    t0 = time.time(); pipe.fit(X_tr_s, y_tr_s); dt = time.time() - t0
    fitted[name] = pipe; timings[name] = dt
    print(f'{name:18s} trained in {dt:6.2f}s')
````

Tree models (`DecisionTree`, `RandomForest`, `XGBoost`) don't actually need the
scaling step, but running every model through the same `preprocessor` keeps one code
path and one persisted object. `max_depth` and `min_samples_leaf` are set
conservatively to limit overfitting; a hyperparameter search wasn't required for this
comparison and could be layered on top later.


## 18. Model comparison

Validation MAE / MSE / RMSE / R² / MAPE and training time for all five models,
ranked by RMSE.


````python
rows = []
for name, pipe in fitted.items():
    s = regression_scores(y_val, pipe.predict(X_val))
    s['model'] = name; s['train_s'] = round(timings[name], 2)
    rows.append(s)
cmp = pd.DataFrame(rows).set_index('model')[['MAE', 'MSE', 'RMSE', 'R2', 'MAPE_%', 'train_s']]
cmp = cmp.sort_values('RMSE')
cmp.round(3)
````

Random Forest posts the lowest validation RMSE and the highest R², with XGBoost
close behind. Linear Regression and Ridge trail both because they can't capture the
area × location × type interactions that drive price; the single Decision Tree lands
in between. Random Forest is the model carried forward to the held-out test set.


## 19. Evaluation

*(on the held-out test set — the split fixed in section 14, untouched during training
or model comparison.)*


````python
BEST = cmp.index[0]          # model with the lowest validation RMSE
print('selected model:', BEST)

final_pipe = fitted[BEST]
test_scores = regression_scores(y_test, final_pipe.predict(X_test))
for k, v in test_scores.items(): print(f'  {k:7s}: {v:,.3f}')

# predicted vs actual (VND scale)
yp = np.expm1(final_pipe.predict(X_test)); yt = y_test_vnd
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
ax[0].scatter(yt, yp, s=6, alpha=0.3); lim = [0, np.percentile(yt, 99)]
ax[0].plot(lim, lim, 'r--'); ax[0].set_xlim(lim); ax[0].set_ylim(lim)
ax[0].set_xlabel('actual Price (million VND)'); ax[0].set_ylabel('predicted'); ax[0].set_title('Predicted vs Actual')
resid = yp - yt
ax[1].scatter(yt, resid, s=6, alpha=0.3); ax[1].axhline(0, color='r', ls='--')
ax[1].set_xlim(lim); ax[1].set_xlabel('actual Price'); ax[1].set_ylabel('residual (pred - actual)')
ax[1].set_title('Residuals vs Actual')
plt.tight_layout(); plt.show()
````

**What each metric means for house-price prediction** (held-out test set):

| Metric | Test value | How to read it for this task |
|---|---|---|
| **MAE** | 10,626 million VND (≈ 10.6 tỷ) | the average listing is off by about 10.6 tỷ in absolute terms — large because the price range spans three orders of magnitude, from a few hundred triệu to hundreds of tỷ, and MAE weights every listing equally. |
| **MSE** | 7.22 × 10⁸ | the mean squared error on the million-VND scale; not read on its own — it exists to define RMSE and to serve as the training loss. |
| **RMSE** | 26,873 million VND | roughly 2.5× the MAE, meaning a minority of listings miss by far more than the typical one — the squared term is dominated by a handful of very large or very cheap outliers (section 20). |
| **R²** | 0.186 | the model explains roughly 19% of the variance in `log1p(Price)` beyond a constant-mean predictor. The mean-only baseline (section 16) scores R² = −0.13, so the model does add real signal, but most price variance stays out of reach from these fields: the pooled area–price correlation is only about 0.02 (section 10, since land and built property pull in opposite directions), and there is no street-level location or year-built data. |
| **MAPE** | 350% | inflated by the many low-priced listings — most sit at 2–5 tỷ — where a modest absolute miss reads as a huge percentage; worth reporting, not worth ranking models by. |

There is no confusion matrix or accuracy figure here — this is regression. The
predicted-vs-actual scatter hugs the diagonal for mid-priced listings, and the
residual spread widens at the high end (heteroscedasticity), which is exactly why the
training target is `log1p(Price)` rather than `Price` itself.


## 20. Error analysis

Where the chosen model breaks down: the worst absolute errors, and the mean
percentage error split by property type and by area band.


````python
err = X_test.copy()
err['actual'] = y_test_vnd
err['pred'] = np.expm1(final_pipe.predict(X_test))
err['abs_err'] = (err['pred'] - err['actual']).abs()
err['pct_err'] = err['abs_err'] / err['actual'] * 100

print('worst 10 absolute errors:')
display(err.sort_values('abs_err', ascending=False).head(10)[
    ['Property Type', 'Province', 'district', 'Area', 'actual', 'pred', 'abs_err', 'pct_err']])

print('\nmean |%err| by property type:')
display(err.groupby('Property Type')['pct_err'].mean().sort_values(ascending=False).round(1))

print('\nmean |%err| by area band:')
err['area_band'] = pd.cut(err['Area'], [0, 50, 100, 200, 500, 10000])
display(err.groupby('area_band')['pct_err'].mean().round(1))
````

The largest errors cluster in a few places: very large or very cheap listings,
where the data thins out at the extremes; sparse types like `Kho, nhà xưởng`; and
districts folded into the `infrequent` bucket. The underlying causes are the missing
street-level location, the missing year-built and legal-status fields, and the
skewed target itself. The straightforward fixes are target-encoding `district`,
adding more attributes, or fitting a model per property type.

**Files produced.** `model/model_pipeline.joblib` holds the fitted preprocessing

## 23. Inference test

Reload the saved artifacts and push one raw listing through them — the exact call
`../api/`'s `POST /predict` makes.


````python
loaded_pipe     = joblib.load('../model/model_pipeline.joblib')
loaded_features = joblib.load('../model/feature_names.joblib')

raw_listing = {          # exactly what the API / mobile client sends (raw, unprocessed)
    'Area': 78.7, 'Width': 4.0, 'Length': np.nan,
    'Bedrooms': 3, 'Bathrooms': 2, 'Floors': 2, 'Alley Width': np.nan,
    'Agent Listing Count': 2,
    'Property Type': 'Nhà riêng', 'Position': 'Đường chính', 'Direction': 'Nam',
    'Road Type': 'Đường nhựa', 'Province': 'an-giang', 'Agent Role': 'Chính chủ',
    'district': 'Rạch Giá',
}
# the engineered missing-flags the client can't know — derive them the same way the notebook did
for c in ['Bedrooms', 'Bathrooms', 'Floors']:
    raw_listing[c + '_missing'] = int(pd.isna(raw_listing.get(c)))

row = pd.DataFrame([raw_listing])[loaded_features]     # right columns, right order
pred_log = float(loaded_pipe.predict(row)[0])
pred_price = float(np.expm1(pred_log))                 # back to million VND

result = {
    'predicted_price': round(pred_price, 2),                       # million VND
    'price_per_m2': round(pred_price / raw_listing['Area'], 4),    # million VND / m²  (same formula as ../api/main.py)
    'currency': 'million VND',
    'model': BEST,
}
print(result)
````

**Contract confirmed.** The reloaded pipeline turns a raw listing dict into
`{predicted_price, price_per_m2, currency, model}` by selecting columns, applying the
saved preprocessing, running the regressor, then `expm1` — nothing gets refitted along
the way. `price_per_m2 = predicted_price / Area` (million VND / m²) uses the same
formula as [`../api/main.py`](../api/main.py), so the notebook and the served model
always agree.
